Combien mange en moyenne un être humain ? source : Food and Agriculture Organization of the United Nations (FAO)

Selon la Food and Agriculture Organization of the United Nations (FAO), les besoins énergétiques moyens d’un être humain sont estimés autour de 2 000 à 2 400 kilocalories par jour. Dans ce projet, nous retenons une valeur moyenne de 2 100 kilocalories par personne et par jour afin d’estimer le nombre théorique de personnes pouvant être nourries (2100 kcal est la valeur de référence utilisée par la FAO et l’OMS pour estimer les besoins énergétiques minimaux moyens d’un adulte).


```python
# Filtrage population sur 2017
population_2017 = population[population["Année"] == 2017]
```

```python
#On commence par faire une jointure entre le data frame population et Dispo_alimentaire afin d'ajouter dans ce dernier la population

df_2017_dispo = dispo_alimentaire.merge(
    population_2017,
    on="Zone",
    how="inner"
)
```

```python
df_2017_dispo.columns
```

```python
#Affichage du nouveau dataframe
df_2017.head(5)
```

```python
#Création de la colonne dispo_kcal avec calcul des kcal disponibles mondialement

# Trouver automatiquement la colonne des kcal/personne/jour
col_kcal = next(c for c in df_2017_dispo.columns if "kcal/personne/jour" in c.lower())

# Création de la colonne dispo_kcal (kcal/an)
df_2017_dispo["dispo_kcal"] = (
    df_2017_dispo[col_kcal]
    * (df_2017_dispo["Population"] * 1000)
    * 365
)

df_2017_dispo[["Zone", "dispo_kcal"]].head()



```

```python
#Calcul du nombre d'humains pouvant être nourris

# 3.2 — Calcul du nombre d'humains pouvant être nourris (alimentation humaine)
besoin_kcal_jour = 2100

# Colonne kcal/personne/jour (tu l'as déjà repérée avec ton next(...))
col_kcal = next(c for c in df_2017_dispo.columns if "kcal/personne/jour" in c.lower())

# Sécurise l'unité de population (si jamais elle est en milliers)
pop_check = df_2017_dispo["Population"].sum()
pop_factor = 1_000 if pop_check < 1e9 else 1  # si somme trop petite -> sûrement en milliers
df_2017_dispo["pop_personnes"] = df_2017_dispo["Population"] * pop_factor

# Total kcal disponibles dans le monde PAR JOUR (somme sur tous produits)
kcal_monde_jour = (df_2017_dispo[col_kcal] * df_2017_dispo["pop_personnes"]).sum()

# Humains théoriquement nourrissables
humains_nourrissables = kcal_monde_jour / besoin_kcal_jour

# Population mondiale réelle (issue du fichier population filtré 2017)
pop_monde_2017 = population_2017["Population"].sum()
pop_monde_2017 = pop_monde_2017 * (1000 if pop_monde_2017 < 1e9 else 1)  # même sécurité

taux_couverture = (humains_nourrissables / pop_monde_2017) * 100

print(f"Kilocalories disponibles dans le monde (par jour) : {kcal_monde_jour/1e12:,.2f} mille milliards kcal".replace(",", " ").replace(".", ","))
print(f"Humains théoriquement nourrissables : {humains_nourrissables/1e9:,.2f} milliards de personnes".replace(",", " ").replace(".", ","))
print(f"Soit environ : {taux_couverture:,.2f} % de la population mondiale".replace(",", " ").replace(".", ","))
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


pop_monde_2017 = pop_monde_2017              # 7_543_798_779
humains_nourrissables = humains_nourrissables # ~9_96e9

# Conversion en milliards
pop_B = pop_monde_2017 / 1e9
nourri_B = humains_nourrissables / 1e9

BLEU = "#2E86C1"

df_plot = pd.DataFrame({
    "Catégorie": [
        "Population mondiale ",
        "Humains théoriquement nourrissables "
    ],
    "Milliards": [pop_B, nourri_B]
})

sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(10, 5), dpi=200)

# --- Barres ---
bars = ax.bar(
    df_plot["Catégorie"],
    df_plot["Milliards"],
    color=["white", BLEU],
    edgecolor="black",
    linewidth=2
)

# --- Titre et axes ---
ax.set_title(
    "Nombre théorique de personnes pouvant être nourries en 2017",
    fontsize=18,
    weight="bold",
    pad=10
)

ax.set_ylabel("Milliards de personnes")
ax.set_xlabel("")

# --- Valeurs au-dessus des barres ---
def fr(x):
    return f"{x:.2f}".replace(".", ",") + " milliards"

for bar, val in zip(bars, df_plot["Milliards"]):
    ax.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height() + 0.15,
        fr(val),
        ha="center",
        va="bottom",
        fontsize=13,
        weight="bold"
    )

# --- Limite Y pour respirer ---
ax.set_ylim(0, max(pop_B, nourri_B) + 1.2)

plt.tight_layout()
plt.show()
```

Synthèse : 
Les résultats montrent que la production alimentaire mondiale aurait été suffisante pour nourrir l’ensemble de la population mondiale en 2017, confirmant que la faim dans le monde est avant tout un problème d’organisation et d’accès aux ressources.

```python
#Transfert des données avec les végétaux dans un nouveau dataframe

# 3.3 — Bloc 1 : données végétales (à partir de df_2017_dispo)
df_veg_2017 = df_2017_dispo[
    df_2017_dispo["Origine"].astype(str).str.strip().str.lower().eq("vegetale")
].copy()

df_veg_2017[["Zone", "Produit", "Origine"]].head()



```

```python
#Calcul du nombre de kcal disponible pour les végétaux

# 3.3 — Bloc 2 : kcal végétales monde / jour

# 1) détecte si "Population" est en milliers ou en personnes (auto)
pop_world_unique = df_veg_2017.groupby("Zone")["Population"].first().sum()
fact_pop = 1000 if pop_world_unique < 1e8 else 1  # ~7,5e6 => milliers ; ~7,5e9 => personnes

# 2) colonne kcal/personne/jour
col_kcal = "Disponibilité alimentaire (Kcal/personne/jour)"

# 3) kcal végétales monde / jour (somme sur tous les produits végétaux)
kcal_veg_monde_jour = (df_veg_2017[col_kcal] * (df_veg_2017["Population"] * fact_pop)).sum()

# 4) population mondiale 2017 (zones uniques) pour info
pop_monde_2017 = pop_world_unique * fact_pop

print(f"Kcal végétales disponibles dans le monde (par jour) : {kcal_veg_monde_jour/1e12:,.2f} mille milliards kcal".replace(",", " ").replace(".", ","))
print(f"Population mondiale (2017) : {pop_monde_2017:,.0f} personnes".replace(",", " "))

```

```python
#Calcul du nombre d'humains pouvant être nourris avec les végétaux

besoin_kcal_jour = 2100

humains_nourrissables_veg = kcal_veg_monde_jour / besoin_kcal_jour

# population mondiale réelle 2017 (celle que tu as déjà validée)
pop_monde_2017 = 7_543_798_779

surplus = humains_nourrissables_veg - pop_monde_2017
taux_couverture = (humains_nourrissables_veg / pop_monde_2017) * 100

print(f"Humains théoriquement nourrissables (végétaux) : {humains_nourrissables_veg/1e9:,.2f} milliards".replace(",", " ").replace(".", ","))
print(f"Surplus théorique : {surplus/1e9:,.2f} milliards".replace(",", " ").replace(".", ","))
print(f"Couverture : {taux_couverture:,.1f} %".replace(",", " ").replace(".", ","))

```
